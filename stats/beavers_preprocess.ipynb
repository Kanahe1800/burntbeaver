```python
# Dropping the columns we do not want
df_relevant = df.drop(columns = dropped_features, errors = "ignore")

#df_relevant.head()

# Checking the remaining relevant columns = 13
print(len(df_relevant.columns))

print()
print("=====================================================")
print()

# Checking for null values in the each of the columns
print(df_relevant.isna().sum())

# After checking for null values count, these columns should be dropped
# because the the majority of the values are null (> 60k). In the case of
# coordinateUncertainty, there are >12k null values with no good basis for imputation

null_columns = ['locality', 'individualCount', 'elevation', 'coordinateUncertaintyInMeters']
df_relevant = df_relevant.drop(columns = null_columns, errors = "ignore")
```

```python
# Double check for null values
print(df_relevant.isna().sum())
print("====================")

# Columns that can be ignored for now: StateProvince, collectionCode
# The eventDate variables are in this format: YY-MM-DD, so we can
# impute the missing day, month, year values with info from eventDate, and vice versa

#print(len(df_relevant))

# First, we convert the eventDate column into Pandas datetime object all at once.
# Since there can be some unique values like '2012-06-15/2012-06-16', the to_datetime()
# function would return NaT, instead of crashing the script
parsed_dates = pd.to_datetime(df_relevant["eventDate"], errors = "coerce")


# Secondly, we would look at the year, month, and day. Then if any of those value
# in the row is NaN, then we would fill it in with the corresponding values from eventDate
df_relevant["year"] = df_relevant["year"].fillna(parsed_dates.dt.year)
df_relevant["month"] = df_relevant["month"].fillna(parsed_dates.dt.month)
df_relevant["day"] = df_relevant["day"].fillna(parsed_dates.dt.day)

  # year, month, date -> EventDate direction
#print(df_relevant.isna().sum())

# Dropping rows with missing year data
df_relevant = df_relevant.dropna(subset = ["year", "decimalLatitude", "decimalLongitude"])
print(df_relevant.isna().sum())

# Dropping columns with year starting before 2015
final_df = df_relevant[df_relevant["year"] >= 2015]
print("==================")
print(len(final_df))
```

```python
# Pipeline to prevent leakage (start to finish)
def dropping_unnecessary_columns(df):
  """ Removes irrelevant columns, columns with all Nan, or many nulls """
  dropped_features = ['gbifID', 'datasetKey', 'kingdom', 'phylum', 'class', 'order',
                    'family', 'genus', 'species', 'infraspecificEpithet', 'taxonRank',
                    'scientificName', 'verbatimScientificName', 'verbatimScientificNameAuthorship',
                    'occurrenceStatus', 'publishingOrgKey', 'coordinatePrecision', 'coordinatePrecision',
                    'elevationAccuracy', 'depth', 'depthAccuracy', 'taxonKey', 'speciesKey',
                    'basisOfRecord', 'institutionCode', 'catalogNumber', 'recordNumber',
                    'identifiedBy', 'dateIdentified', 'license', 'rightsHolder',
                    'recordedBy', 'typeStatus', 'establishmentMeans', 'lastInterpreted',
                    'mediaType', 'issue', 'occurrenceID', 'locality', 'individualCount',
                    'elevation', 'coordinateUncertaintyInMeters']
  return df.drop(columns = dropped_features, errors = "ignore")

def impute_missing_data(df):
  """ Extract year, month, day from eventDate using the to_datetime() func """
  parsed_dates = pd.to_datetime(df["eventDate"], format = "%Y-%m-%d", errors = "coerce")

  # .assign() is the pipeline friendly way to create or update columns
  return df.assign(
      year = df["year"].fillna(parsed_dates.dt.year),
      month = df["month"].fillna(parsed_dates.dt.month),
      day = df["day"].fillna(parsed_dates.dt.day)
  )

def filter_timeline(df, start_year):
  """ Filter the dataset to only include rows after 2015 """
  # Create a boolean mask and return the filtered dataframe
  return df[df["year"] >= start_year]

# Pipeline execution
clean_beavers_df = (
    df
    .pipe(dropping_unnecessary_columns)
    .pipe(impute_missing_data)
    .pipe(filter_timeline, 2015)
    .dropna(subset = ["year", "decimalLatitude", "decimalLongitude"])
)

clean_beavers_df.head()
print(len(clean_beavers_df))
clean_beavers_df.head()
#clean_beavers_df.to_csv("Beavers North american info from 2015 till now clean", index = False)
```
